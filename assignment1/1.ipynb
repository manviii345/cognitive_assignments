```python
n = int(input("Enter n: "))
sum = 0

for num in range(2, n + 1):
    count = 0
    for i in range(1, num + 1):
        if num % i == 0:
            count += 1

    if count == 2:
        sum += num

print("Sum of prime numbers =", sum)
```

```python
def addOdd(n):
    sum = 0
    for i in range(1,n+1):
        if i%2 == 1:
            sum += i
    return sum
n = int(input("Enter the number:"))
print("Sum of all odd numbers till",n,"is",addOdd(n))
```

```python
def prime_sum(n):
    total = 0

    for num in range(2, n + 1):
        count = 0

        for i in range(1, num + 1):
            if num % i == 0:
                count += 1

        if count == 2:
            total += num

    return total


n = int(input("Enter n: "))
print("Sum of prime numbers =", prime_sum(n))
```
