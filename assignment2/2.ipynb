```python
import random
random.seed(1024170367)
L = [random.randint(100,900) for i in range(100)]
print(L)
```

```python
cnt = 0;
for val in L:
    if val%2 == 1:
        cnt += 1
print("No. of odd numbers: ", cnt)
```

```python
cnt = 0;
for val in L:
    if val%2 == 0:
        cnt += 1
print("No. of even numbers: ", cnt)
```

```python
def checkPrime(n):
    if n<2:
        return True
    for i in range(2,int(n**0.5)+1):
        if n%i == 0:
            return False
    return True
primes = [num for num in L if checkPrime(num)]
print("Count of primes:", len(primes))
print("Primes:", primes)
```

```python
freq = {}
for num in L:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
most_freq_num = 0;
cnt = 0;
for key in freq:
    if freq[key] > cnt:
        cnt = freq[key]
        most_freq_num = key
print("Most frequent elemet:",most_freq_num)
print("Count of most frequent element:",cnt)
```

```python
roll = '10241703'
A = {int(digit)*7 for digit in roll}
B = {int(digit)*9 for digit in roll}
print(A)
print(B)
```

```python
union_set = A.union(B)
print(union_set)
```

```python
intersection_set = A.intersection(B)
print(intersection_set)
```

```python
A_diff_B = A.difference(B)
print("A-B:",A_diff_B)
B_diff_A = B.difference(A)
print("B-A:",B_diff_A)
# difference() gives elements present in one set but not the other,
# while symmetric_difference() gives elements unique to BOTH sets.
```

```python
sym_diff = A.symmetric_difference(B)
print("Symmetric difference of A and B:", sym_diff)
```

```python
is_sub = A.issubset(B)
is_super = A.issuperset(B)
print("Is A a subset of B?:", is_sub)
print("Is B a superset of A?:", is_super)
```

```python
x = int(input("Enter a value X to remove from set A: "))
A.discard(x)
print("Updated Set A after discard:", A)
#discard() removes an element without raising an error if it doesn't exist, whereas remove() throws a KeyError if the element is missing.
```
