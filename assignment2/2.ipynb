```python
scores = tuple(L[i] for i in range(8))
print(scores)
```

```python
highest_score = max(scores)
print("Highest score: ", highest_score)
for i in range(8):
    if(scores[i] == highest_score):
        index = i
        break
print(index)

lowest_score = min(scores)
print("Lowest score: ", lowest_score)
cnt = 0
for score in scores:
    if score == lowest_score:
        cnt += 1
print(cnt)
```

```python
temp = list(scores)
temp.reverse()
scores = tuple(temp)
print(scores)
# The tuple cannot be reversed because we cannot change the value of any element inside a tuple
```

```python
index = -1
n = len(scores)
score = int(input("Enter the element to find: "))
for i in range(n):
    if scores[i] == score:
        index = i
        break
if index == -1:
    print("Element is not present in the tuple")
else:
    print("First occurrence of element found at index ", index)
```

```python
try:
    scores[0] = 60
except TypeError:
    print("Cannot change values in an immutable data structure which is tuple. If it was list, then no error would be raised as list is a mutable data structure")
```

```python
first_score, second_score, *remaining_scores = scores
print(first_score)
print(second_score)
print(remaining_scores)
```

```python
import random
random.seed(1024170367)
L = [random.randint(100,900) for i in range(100)]
print(L)
```

```python
cnt = 0;
for val in L:
    if val%2 == 1:
        cnt += 1
print("No. of odd numbers: ", cnt)
```

```python
cnt = 0;
for val in L:
    if val%2 == 0:
        cnt += 1
print("No. of even numbers: ", cnt)
```

```python
def checkPrime(n):
    if n<2:
        return True
    for i in range(2,int(n**0.5)+1):
        if n%i == 0:
            return False
    return True
primes = [num for num in L if checkPrime(num)]
print("Count of primes:", len(primes))
print("Primes:", primes)
```

```python
freq = {}
for num in L:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
most_freq_num = 0;
cnt = 0;
for key in freq:
    if freq[key] > cnt:
        cnt = freq[key]
        most_freq_num = key
print("Most frequent elemet:",most_freq_num)
print("Count of most frequent element:",cnt)
```

```python
roll = '10241703'
A = {int(digit)*7 for digit in roll}
B = {int(digit)*9 for digit in roll}
print(A)
print(B)
```

```python
union_set = A.union(B)
print(union_set)
```

```python
intersection_set = A.intersection(B)
print(intersection_set)
```

```python

print("A - B =", A.difference(B))
```

```python
print("B - A =", B.difference(A))
# difference() gives elements present in one set but not the other,
# while symmetric_difference() gives elements unique to BOTH sets.
```
