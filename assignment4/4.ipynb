Question 2

```python
def score_query(query, df):
    query_words = query.lower().split()
    results = []

    for index, row in df.iterrows():
        text = (
            row["question"] + " " +
            row["answer"] + " " +
            row["keywords"]
        ).lower()

        score = 0

        for word in query_words:
            if word in text:
                score += 1

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    results.sort(key=lambda x: x["score"], reverse=True)

    return results


query = input("Enter your query: ")

results = score_query(query, df)

for result in results:
    print(result)
```

```python
def score_query_with_ties(query, df):
    query_words = query.lower().split()
    results = []

    for i, row in df.iterrows():

        text = (
            str(row["question"]) + " " +
            str(row["answer"]) + " " +
            str(row["keywords"])
        ).lower()

        score = 0

        for word in query_words:
            if word in text:
                score += 1

        if score > 0:
            results.append((i, score))

    # No matches
    if len(results) == 0:
        print("No matching entries.")
        return

    # Find highest score
    highest_score = max(score for i, score in results)

    print("Highest score:", highest_score)
    print("Matching entries:")

    # Print ALL entries having highest score
    for i, score in results:
        if score == highest_score:
            print("\nQuestion:", df.loc[i, "question"])
            print("Answer:", df.loc[i, "answer"])
            print("Category:", df.loc[i, "category"])
            print("Score:", score)


# -------- Q6 DEMONSTRATION --------

print("----- QUERY: fee -----")
score_query_with_ties("fee", df)

print("\n----- QUERY: password -----")
score_query_with_ties("password", df)
```
